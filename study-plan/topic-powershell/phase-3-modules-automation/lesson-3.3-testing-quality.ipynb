## Concepts

Testing tells you whether the script still behaves the way you expect after a change. For PowerShell, that usually means Pester tests for behavior and PSScriptAnalyzer for style and safety rules.

Quality checks are most useful when they run early, before a change gets merged or deployed.

Reference: [PowerShell Documentation](https://learn.microsoft.com/en-us/powershell/)

```python
from pathlib import Path
import shutil
import subprocess
import textwrap

lesson_dir = Path('study-plan/topic-powershell/phase-3-modules-automation')
tests_dir = lesson_dir / 'tests'
artifact_dir = lesson_dir / 'artifacts'
tests_dir.mkdir(parents=True, exist_ok=True)
artifact_dir.mkdir(parents=True, exist_ok=True)

def run_powershell(script_text: str, script_name: str) -> int:
    script_path = lesson_dir / script_name
    script_path.write_text(script_text, encoding='utf-8')
    executable = shutil.which('pwsh') or shutil.which('powershell')
    if executable:
        result = subprocess.run([executable, '-NoLogo', '-NoProfile', '-ExecutionPolicy', 'Bypass', '-File', str(script_path)], capture_output=True, text=True)
        print(result.stdout)
        if result.returncode != 0:
            print(result.stderr)
        return result.returncode
    print(f'{script_path.name} written. Run it in PowerShell to see the live output.')
    return 0
```

## Code-Along: Step 1 - Write a Pester test file

This step writes a test file that checks the module function returns the shape you expect. The tests are intentionally small so the focus stays on the testing workflow.

```python
test_path = tests_dir / 'PowerShellDevOpsToolkit.Tests.ps1'
test_content = textwrap.dedent(r'''
Describe 'PowerShellDevOpsToolkit' {
    It 'returns a status object' {
        $result = [pscustomobject]@{ ModuleName = 'PowerShellDevOpsToolkit'; Status = 'Ready' }
        $result.ModuleName | Should -Be 'PowerShellDevOpsToolkit'
        $result.Status | Should -Be 'Ready'
    }
}
''')
test_path.write_text(test_content, encoding='utf-8')
print(test_path.read_text(encoding='utf-8'))
```

## Code-Along: Step 2 - Add analyzer settings

Analyzer settings let you choose which rules matter for this toolkit. That keeps the check focused on the standards your team actually wants to enforce.

```python
settings_path = lesson_dir / 'PSScriptAnalyzerSettings.psd1'
settings_content = textwrap.dedent(r'''
@{
    IncludeRules = @('PSAvoidUsingWriteHost', 'PSUseApprovedVerbs')
    ExcludeRules = @()
}
''')
settings_path.write_text(settings_content, encoding='utf-8')
print(settings_path.read_text(encoding='utf-8'))
```
