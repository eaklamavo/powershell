## Concepts

A **module** is a package of commands that PowerShell can load together. Instead of copy/pasting functions between scripts, you place them in one module file and import that module where needed.

A **manifest** describes the module: its name, version, author, and exported functions. That makes the module easier to install, version, and share.

Reference: [PowerShell Documentation](https://learn.microsoft.com/en-us/powershell/)

```python
from pathlib import Path
import shutil
import subprocess
import textwrap

lesson_dir = Path('study-plan/topic-powershell/phase-3-modules-automation')
src_dir = lesson_dir / 'src/PowerShellDevOpsToolkit'
artifact_dir = lesson_dir / 'artifacts'
src_dir.mkdir(parents=True, exist_ok=True)
artifact_dir.mkdir(parents=True, exist_ok=True)

def run_powershell(script_text: str, script_name: str) -> int:
    script_path = lesson_dir / script_name
    script_path.write_text(script_text, encoding='utf-8')
    executable = shutil.which('pwsh') or shutil.which('powershell')
    if executable:
        result = subprocess.run([executable, '-NoLogo', '-NoProfile', '-ExecutionPolicy', 'Bypass', '-File', str(script_path)], capture_output=True, text=True)
        print(result.stdout)
        if result.returncode != 0:
            print(result.stderr)
        return result.returncode
    print(f'{script_path.name} written. Run it in PowerShell to see the live output.')
    return 0
```

## Code-Along: Step 1 - Create a reusable module file

This step creates a small module with two functions that belong together. The names are intentionally generic so the module can grow in later lessons and in the capstone.

```python
module_path = src_dir / 'PowerShellDevOpsToolkit.psm1'
module_content = textwrap.dedent(r'''
function Get-ToolkitStatus {
    [CmdletBinding()]
    param()

    [pscustomobject]@{
        ModuleName = 'PowerShellDevOpsToolkit'
        Status     = 'Ready'
        Timestamp  = Get-Date
    }
}

function New-ToolkitTag {
    [CmdletBinding()]
    param(
        [Parameter(Mandatory)]
        [string]$Name,

        [Parameter(Mandatory)]
        [string]$Value
    )

    @{ Name = $Name; Value = $Value }
}

Export-ModuleMember -Function Get-ToolkitStatus, New-ToolkitTag
''')
module_path.write_text(module_content, encoding='utf-8')
print(module_path.read_text(encoding='utf-8'))
```

## Code-Along: Step 2 - Add a manifest

A manifest is the metadata layer for a module. It helps PowerShell know what the module exports and what version it is.

```python
manifest_path = src_dir / 'PowerShellDevOpsToolkit.psd1'
manifest_content = textwrap.dedent(r'''
@{
    RootModule        = 'PowerShellDevOpsToolkit.psm1'
    ModuleVersion     = '0.1.0'
    GUID              = '11111111-1111-1111-1111-111111111111'
    Author            = 'Study Plan'
    CompanyName       = 'Learning Lab'
    PowerShellVersion = '7.0'
    FunctionsToExport = @('Get-ToolkitStatus', 'New-ToolkitTag')
}
''')
manifest_path.write_text(manifest_content, encoding='utf-8')
print(manifest_path.read_text(encoding='utf-8'))
```
