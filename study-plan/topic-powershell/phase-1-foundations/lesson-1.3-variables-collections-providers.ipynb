## Concepts

A **variable** stores a value. An **array** stores an ordered list of values. A **hash table** stores key/value pairs, which is useful when you want names like `Name` or `Region` attached to data.

PowerShell also exposes **providers**, which let the shell treat different data stores like navigable drives. The file system is the most common provider, but the same idea helps you reason about other data sources later.

Reference: [PowerShell Documentation](https://learn.microsoft.com/en-us/powershell/)

```python
from pathlib import Path
import json
import shutil
import subprocess
import textwrap

lesson_dir = Path('study-plan/topic-powershell/phase-1-foundations')
artifact_dir = lesson_dir / 'artifacts'
artifact_dir.mkdir(parents=True, exist_ok=True)

def run_powershell(script_text: str, script_name: str) -> int:
    script_path = lesson_dir / script_name
    script_path.write_text(script_text, encoding='utf-8')
    executable = shutil.which('pwsh') or shutil.which('powershell')
    if executable:
        result = subprocess.run([executable, '-NoLogo', '-NoProfile', '-ExecutionPolicy', 'Bypass', '-File', str(script_path)], capture_output=True, text=True)
        print(result.stdout)
        if result.returncode != 0:
            print(result.stderr)
        return result.returncode
    print(f'{script_path.name} written. Run it in PowerShell to see the live output.')
    return 0
```

## Code-Along: Step 1 - Build a small inventory

This step creates a hash table that describes a simple DevOps asset inventory. The point is not the data itself; the point is to practice the shape of a real automation input.

```python
script = textwrap.dedent(r'''
$inventory = @{
    Project = 'PowerShell DevOps Study Plan'
    Region  = 'eastus'
    Roles   = @('build-agent', 'reporting', 'automation')
}

$inventory | ConvertTo-Json -Depth 3
''')
run_powershell(script, 'lesson-1.3-inventory.ps1')
```

## Code-Along: Step 2 - Work with paths and folders

PowerShell makes file-system work safer when you build paths programmatically. `Join-Path` avoids hard-coding separators, and `Test-Path` lets you check whether a file or folder already exists before creating it.

```python
script = textwrap.dedent(r'''
$root = Join-Path $PSScriptRoot 'artifacts/inventory-demo'
if (-not (Test-Path $root)) { New-Item -ItemType Directory -Path $root | Out-Null }
Set-Content -Path (Join-Path $root 'roles.txt') -Value 'build-agent','reporting','automation'
Get-ChildItem -Path $root | Select-Object Name, Length
''')
run_powershell(script, 'lesson-1.3-paths-and-folders.ps1')
```
