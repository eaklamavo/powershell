## Concepts

PowerShell is a shell and scripting language designed around **objects**, not just plain text. That means commands return structured data with properties you can inspect, filter, and pass to other commands.

A **cmdlet** is a built-in PowerShell command such as `Get-Help` or `Get-ChildItem`. A **pipeline** is the `|` operator that sends output from one command into the next command.

Reference: [PowerShell Documentation](https://learn.microsoft.com/en-us/powershell/)

```python
from pathlib import Path
import shutil
import subprocess
import textwrap

lesson_dir = Path('study-plan/topic-powershell/phase-1-foundations')
artifact_dir = lesson_dir / 'artifacts'
artifact_dir.mkdir(parents=True, exist_ok=True)

def run_powershell(script_text: str, script_name: str) -> int:
    script_path = lesson_dir / script_name
    script_path.write_text(script_text, encoding='utf-8')
    executable = shutil.which('pwsh') or shutil.which('powershell')
    if executable:
        result = subprocess.run([executable, '-NoLogo', '-NoProfile', '-ExecutionPolicy', 'Bypass', '-File', str(script_path)], capture_output=True, text=True)
        print(result.stdout)
        if result.returncode != 0:
            print(result.stderr)
        return result.returncode
    print(f'{script_path.name} written. Run it in PowerShell to see the live output.')
    return 0
```

## Code-Along: Step 1 - Inspect your shell

This step writes a small script that prints the PowerShell version table and a few useful environment values. The goal is to see that PowerShell exposes structured state that scripts can inspect directly.

```python
script = textwrap.dedent(r'''
$PSVersionTable | Format-List
Write-Host ''
Write-Host "Home folder: $HOME"
Write-Host "Is Windows: $IsWindows"
Get-Command Get-Help, Get-ChildItem | Select-Object Name, CommandType
''')
run_powershell(script, 'lesson-1.1-first-commands.ps1')
```

## Code-Along: Step 2 - Discover help and commands

A major PowerShell habit is to let the shell teach you. `Get-Help` explains commands, and `Get-Command` lets you search by verb, noun, or module. This is often faster than searching the web while you are already in a terminal session.

```python
script = textwrap.dedent(r'''
Get-Help Get-Process | Select-Object Name, Synopsis
Write-Host ''
Get-Command -Verb Get | Select-Object -First 10 Name, Source
''')
run_powershell(script, 'lesson-1.1-help-discovery.ps1')
```
